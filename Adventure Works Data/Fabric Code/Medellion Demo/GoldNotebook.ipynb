```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
```

```python
# Read the SilverLakehouse managed Delta table
silver_customers = spark.read.format("delta").load("abfss://NitinFabricWS@onelake.dfs.fabric.microsoft.com/SilverLakehouse.Lakehouse/Tables/dbo/silver_customers")
display(silver_customers.limit(10))
```

```python
df_dim_customers = silver_customers.select(
    "CustomerKey", "FullName", "Gender", "BirthDate", 
    "MaritalStatus", "EmailAddress", "AnnualIncome", "EducationLevel", 
    "Occupation", "HomeOwner", "domain","TotalChildren"
).dropDuplicates(["CustomerKey"])
df_dim_customers.createOrReplaceTempView("dim_customers_view")
```

```python
%%sql
CREATE TABLE GoldWarehouse.gold.dim_customers
AS
SELECT * FROM dim_customers_view;
```
