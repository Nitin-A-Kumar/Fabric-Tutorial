# Reading Bronze Lakehouse Data

**Reading Customer Data and applying transformation**

```python
# infershema detect datatype
df_customer = spark.read.format("csv")\
.option("header",True)\
.option("inferschema",True)\
.load("abfss://NitinFabricWS@onelake.dfs.fabric.microsoft.com/bronzeLakehouse.Lakehouse/Files/BronzeRawDataBronze_Customers")
display(df_customer.limit(10))
```

```python
df_customer.printSchema()
```

```python
#withcolumn to create new column
df_customer = df_customer.withColumn("Full Name",concat(col("Prefix"),lit(" "),col("FirstName"),lit(" "),col("LastName"),lit(" ")))
display(df_customer.limit(10))
```

```python
df_customer = df_customer.drop("Fullname")
display(df_customer.limit(10))
```

```python

df_customer = df_customer.drop("Fullname")
df_customer = df_customer.withColumnRenamed("Full Name", "FullName")
display(df_customer.limit(10))

```

```python
df_customer = df_customer.withColumn(
    "FullName",
    initcap(col("FullName"))
)
display(df_customer.limit(10))
```

```python
df_customer = df_customer.withColumn("domain",split("EmailAddress","@")[1])
display(df_customer.limit(10))
```

# Wrtie the data

```python
df_customer.write \
    .format("delta")\
    .mode("overwrite")\
    .saveAsTable("Silver_Customers")
```

```python
df_customer.write \
    .mode("overwrite") \
    .option("header", "true") \
    .csv("Files/Silver_Customer")
```

* Reading Customer Data and applying transformation

```python

df_sales_2016 = spark.read.format("csv")\
.option("header",True)\
.option("inferschema",True)\
.load("abfss://NitinFabricWS@onelake.dfs.fabric.microsoft.com/bronzeLakehouse.Lakehouse/Files/BronzeRawDataBronze_Sales_2016")
display(df_sales_2016.limit(10))


df_sales_2017 = spark.read.format("csv")\
.option("header",True)\
.option("inferschema",True)\
.load("abfss://NitinFabricWS@onelake.dfs.fabric.microsoft.com/bronzeLakehouse.Lakehouse/Files/BronzeRawDataBronze_Sales_2017")
display(df_sales_2017.limit(10))
```

```python
df_sales = df_sales_2016.union(df_sales_2017)
display(df_sales.limit(10))
```


* Write sales data

```python
df_sales.write \
    .format("delta") \
    .mode("overwrite")\
     .saveAsTable("Silver_Sales")
```

```python
df_sales.write \
    .mode("overwrite") \
    .option("header", True) \
    .csv("Files/Silver_Sales")
```
